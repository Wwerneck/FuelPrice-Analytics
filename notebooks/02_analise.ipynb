# Análise de preços
Indicadores reproduzíveis sobre a camada processada.

```python
import pandas as pd
df = pd.read_parquet('../data/processed/fuel_prices.parquet')
display(df.groupby('produto').preco_venda.mean().sort_values(ascending=False))
display(df.groupby('uf').preco_venda.agg(['mean','min','max','std']).sort_values('mean', ascending=False))
display(df.groupby(['ano_mes','produto'], as_index=False).preco_venda.mean())
```
