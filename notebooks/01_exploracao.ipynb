# Exploração dos dados ANP
Execute o pipeline antes deste notebook. A análise usa somente a camada processada.

```python
import pandas as pd
df = pd.read_parquet('../data/processed/fuel_prices.parquet')
display(df.head(), df.tail())
df.info()
print(df.shape, df.columns.tolist(), df.dtypes, df.duplicated().sum())
display(df.describe(include='all'), df.isna().sum())
```

```python
for column in ['produto', 'regiao', 'uf', 'municipio', 'ano_mes']:
    display(df[column].value_counts(dropna=False).head(30))
display(df.groupby('produto').preco_venda.describe())
display(df.loc[df.outlier].sort_values('z_score', key=abs, ascending=False).head(30))
```
